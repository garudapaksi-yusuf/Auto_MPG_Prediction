### Conclusion & Recommendation

The Auto MPG dataset contains variables relevant to predicting city-cycle fuel consumption. Data exploration suggests that point predictions are better made at the midrange of the (probability) distribution of mpg. Median absolute error is used to evaluate prediction performance; it is also directly interpretable as the average magnitude of prediction errors. These findings motivate the use of histogram-based gradient boosting as a simple implementation capable of handling issues in the data

* As an ensemble model built from several decision trees, gradient boosting also offers tunable hyperparameters beyond those used here, such as `min_samples_leaf`, `max_leaf_nodes`, and a feature to implement [monotonic constraints](https://scikit-learn.org/stable/modules/ensemble.html#monotonic-constraints)
    * Future development of this project may have an interest in exploring these opportunities to improve the model further.
    * Alternatively, current model can be refined by extending the existing hyperparameter grid in the desired direction or conducting a more granular search beyond its current range.
* Baseline performance of `histgradientboostingregressor` provided by the `6`-fold cross validation procedure is $\text{MedAE} = 1.46 ± 0.31$
    * Because the number of folds was chosen arbitrarily based on dataset's size, [Sensitivity analysis](https://en.wikipedia.org/wiki/Sensitivity_analysis) for different values of $k$ could be performed. Larger values of $k$ are expected to yield a less noisy, more precise estimate of the model’s performance.
    * An alternate approach is to use other cross-validation iterators such as [Repeated or Leave-One-Out](https://scikit-learn.org/stable/modules/cross_validation.html#cross-validation-iterators-for-i-i-d-data), at the cost of higher `fit_time`.
    * One possible source of prediction errors is actual `mpg` values at the upper end of the distribution; additional data in that range may help reduce the errors.
* In general, this project may be classified as [observational study](https://en.wikipedia.org/wiki/Observational_study) towards investigating the Auto MPG dataset without controlling any of the variable inside the data.
    * This kind of studies, for lacking an assignment mechanism and combined with multicollinearity issue detected in aforementioned sections, naturally present difficulties for inferential or [interpretability](https://scikit-learn.org/stable/inspection.html) analysis —to understand the predictions and what affects them but the insight may be unreliable and misleading.
    * Furthermore, informations provided here may also be used to develop an in-depth use-case scenario to demonstrate the practical application of this project.
